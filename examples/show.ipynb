# Temple plot demo gallery

Illustrative synthetic datasets for choosing a README/GIF example.

```python
from temple_plot import TemplePlot
from demo_data import (
    DEMO_CLASSES,
    TERMINI_DEMO_CLASSES,
    TERMINI_DEMO_CLASSES_ALT,
    CONTINENT_AGE_CLASSES,
    INCOME_GROUP_AGE_CLASSES,
    LAND_USE_CLASSES,
    HOUSEHOLD_BUDGET_CLASSES,
)
```

## Generic composition demo

```python
fig = TemplePlot(
    DEMO_CLASSES,
    title="Evidence composition by dataset",
    orientation="landscape",
    master_legend_title="Datasets",
    composition_legend_title="Evidence levels",
    composition_axis_title="Share within dataset",
    master_axis_title="Total evidence",
).create_figure()
fig.show()
```

## Termini specificity demo

```python
termini_fig = TemplePlot(
    TERMINI_DEMO_CLASSES,
    title="Termini specificity composition",
    orientation="landscape",
    master_legend_title="Termini classes",
    composition_legend_title="N-terminus states",
    composition_axis_title="Share within termini class",
    master_axis_title="Total identifications",
).create_figure()
termini_fig.show()

termini_portrait_fig = TemplePlot(
    TERMINI_DEMO_CLASSES_ALT,
    title="Termini specificity composition, portrait",
    orientation="portrait",
    width=760,
    height=900,
    master_legend_title="Termini classes",
    composition_legend_title="N-terminus states",
    composition_axis_title="Share within termini class",
    master_axis_title="Total identifications",
).create_figure()
termini_portrait_fig.show()
```
